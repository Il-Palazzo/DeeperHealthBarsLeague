```python
import torch
from ultralytics import YOLO
import multiprocessing

if __name__ == '__main__':
    # --- 1. HARDWARE CHECK ---
    print("========================================")
    print(f"CUDA Available: {torch.cuda.is_available()}")
    if torch.cuda.is_available():
        print(f"GPU Detected: {torch.cuda.get_device_name(0)}")
    print("========================================\n")

    # To avoid Windows multiprocessing freezing
    multiprocessing.freeze_support() 

    # --- 2. LOAD PRE-TRAINED MODEL ---
    # Use .pt to leverage transfer learning (much faster convergence)
    model = YOLO('yolov8n.yaml')  # Extra Large

    # --- 3. START TRAINING WITH UI-OPTIMIZED HYPERPARAMETERS ---
    model.train(
        # Setup
        data=r'C:\Users\USER\Documents\GitHub\DeeperHealthBarsLeague\LolHealth-Bar\yolo_training_data\data.yaml',
        epochs=100,
        imgsz=640,
        device=0,
        
        # Hardware Optimization for RTX 3070 (8GB VRAM)
        batch=32,    # 32 is perfect for YOLO Nano at 640p on 8GB VRAM
        workers=8,   # Speeds up data loading to keep the GPU fed
        
        # --- LEARNING DYNAMICS ---
        lr0=0.001,
        lrf=0.01,
        momentum=0.937,
        weight_decay=0.0005,
        warmup_epochs=3.0,
        
        # --- UI-SPECIFIC AUGMENTATIONS ---
        # DISABLE spatial augmentations that break UI geometry
        fliplr=0.0,       # NEVER mirror UI elements
        flipud=0.0,       # NEVER flip upside down
        mosaic=0.0,       # Don't chop up UI screens
        mixup=0.0,        # Don't blend screens
        degrees=0.0,      # We don't want barrel-roll rotations
        perspective=0.0,  # We already handled this perfectly in the dataset generator
        scale=0.1,        # Allow tiny zoom adjustments
        translate=0.1,    # Allow slight panning
        
        # ALLOW color augmentations (helps handle monitor differences/glare)
        hsv_h=0.015,
        hsv_s=0.5,
        hsv_v=0.4
    )
```
